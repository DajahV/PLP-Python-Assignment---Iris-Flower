# Task 3: Data Visualization

```python

# Pairplot
sns.pairplot(iris, hue='flower_name')
plt.title('Pairplot of Iris Dataset')
plt.show()
```

```python

# Boxplot
plt.figure(figsize=(10, 6))
sns.boxplot(x='flower_name', y='sepal length (cm)', data=iris)
plt.title('Boxplot of Sepal Length by Species')
plt.show()
```

```python
# Create a histogram for each feature in the iris dataset
iris.hist(bins=20, figsize=(15, 10), layout=(2, 2))
plt.suptitle('Histograms of Iris Dataset Features')
plt.show()
```

```python
plt.scatter(iris['sepal length (cm)'],iris['sepal width (cm)'])
plt.scatter(iris['sepal length (cm)'],iris['sepal width (cm)'])
plt.xlabel('length')
plt.ylabel('width')
```
